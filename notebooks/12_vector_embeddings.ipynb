## 1. Create text embeddings

An **embedding model** maps each input text to a fixed-length list of numbers (a vector). The vector encodes learned relationships useful for semantic comparison; its individual numbers are not human-readable definitions. `embed_documents` creates one vector per text. This hosted OpenAI call needs `OPENAI_API_KEY` and may incur charges.

**Checkpoint:** How many vectors should `embed_documents` return for five inputs?

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
documents = [
"Machine learning is a subset of artificial intelligence",
"Deep learning uses neural networks with multiple layers",
"Python is a programming language popular for data science",
"India won the cricket world cup in 2011",
"AI models can understand and generate human language"
]
```

```python
from langchain_openai import OpenAIEmbeddings
```

```python
embeddings=OpenAIEmbeddings(model="text-embedding-3-small")
```

```python
vector=embeddings.embed_documents(documents)
```

```python
len(vector)
```

## 2. Store vectors and retrieve similar text

Chroma stores the source text with its embedding. A similarity search ranks passages by embedding proximity to the query.

A **vector store** keeps source texts, their embeddings, and often metadata. At query time it embeds the query compatibly and ranks stored vectors by a distance/similarity measure; `k=2` requests two matches. The results are candidates, not proof that the answer is supported. This lesson follows [Lesson 11](11_rag_splitters.ipynb) and prepares for the full RAG workflow in [Lesson 13](13_rag_based_pdf_qna_bot.ipynb).

**Checkpoint:** Why should the original text be stored alongside the vector?

```python
from langchain_chroma import Chroma
```

```python
vectore_store=Chroma.from_texts(
    texts=documents,
    embedding=embeddings,
    persist_directory="./vector_db"
)
```

```python
query="Data Science"
result=vectore_store.similarity_search(query=query,k=2)
result
```
