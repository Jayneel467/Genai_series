## 1. Connect to OpenAI

Create a LangChain chat model, send a prompt, and inspect the returned message.

A **chat-model integration** adapts a provider model to LangChain’s message interface. `ChatOpenAI` creates a client; `.invoke(...)` sends one request and returns an AI message. The printed `.content` is the generated text. The provider and selected model determine credentials, capabilities, availability, and cost.

**Checkpoint:** Which part is the provider integration, and which part is the particular model name?

```python
from langchain_openai import ChatOpenAI
llm=ChatOpenAI(model="gpt-4o")
respose_fromgpt=llm.invoke("Explain me 2 main formulas of Integration")
print(respose_fromgpt.content)
```

## 2. Connect to Google Gemini

Compare the provider-specific integration while keeping the same chat-model workflow.

This uses the same high-level pattern with Google’s integration, but a provider-specific class, key, and model identifier. A common LangChain interface makes examples easier to compare; it does not make providers identical.

**Checkpoint:** What must you change when switching providers?

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm=ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
res=llm.invoke("How to find the largest number from a list in Python?")
print(res.content)
```

## 4. Send role-based messages

A system message sets behavior; a user message supplies the task. The example starts a new request rather than retaining conversation history.

A **system message** provides instructions, a **user message** asks a question, and an **assistant/AI message** represents a previous response. The list order is important. This example is one request containing multiple turns; it is not persistent memory. See [Lesson 03](03_basic_memory_with_langchain.ipynb).

**Checkpoint:** What changes if the earlier assistant message is omitted?

```python
from langchain_openai import ChatOpenAI
llm=ChatOpenAI(model="gpt-4o")
```

```python
# Each tuple contains: (message role, message content).
# "system" sets the assistant's behavior and expertise.
# "user" contains the question or task for the model.
# "ai" represents a previous response from the assistant.
# Example: ("ai", "Here is an earlier assistant response.")
prompts=[
    ("system", "You are a Java developer."),
    ("user", "How do I sort an array in Java?"),
    ("ai", "Here is an example response.")
]

# To include conversation history, add an "ai" message before the next "user" message.
# prompts.insert(2, ("ai", "I can help you with Java arrays."))

# Send the messages to the model in one request.
res=llm.invoke(prompts)
print(res.content)
```
