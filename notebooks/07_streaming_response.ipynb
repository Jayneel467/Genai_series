## 2. Stream a response from OpenAI

A **stream** yields response chunks as they arrive instead of waiting for the complete message. This can improve perceived responsiveness, but it is still a model request and does not reduce the need to handle errors or cost. The loop prints each chunk’s text as it arrives.

**Checkpoint:** What does the loop receive on each iteration?

```python
from langchain_openai import ChatOpenAI
```

```python
llm=ChatOpenAI(model="gpt-4o-mini",streaming=True)
```

```python
question="can u explain me gen ai"
```

```python
res=llm.stream(question)
```

```python
for chunk in res:
    print(chunk.content,end="")
```

## 3. Stream a response from Groq

The second example uses the same incremental-consumption pattern with Groq. Chunk boundaries are transport details; they are not guaranteed to align with words or sentences. Compare the setup and iteration with the OpenAI version.

**Checkpoint:** Why should an app not assume each streamed chunk is a complete sentence?

```python
from langchain_groq import ChatGroq
llm=ChatGroq(model="openai/gpt-oss-20b")
```

```python
question="can u explain me gen ai"
response=llm.stream(question)
for chunks in response:
    print(chunks.content,end="")
```
