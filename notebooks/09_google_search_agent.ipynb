## 1. Configure search and model access

Google Serper supplies web-search results; Groq supplies the chat model that interprets them. These are separate services with separate credentials. The search wrapper is a tool, not a source of guaranteed or exhaustive truth.

**Checkpoint:** Which key is for search, and which is for the language model?

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langgraph.checkpoint.memory import MemorySaver
```

## 2. Create an agent with web search

The search function is provided to the agent as a **tool**. When the question may need current information, the model can choose to call it, receive results, and write an answer. Search results can be stale, incomplete, or misleading; check the source pages.

**Checkpoint:** What evidence should you inspect before trusting a search-based answer?

```python
llm=ChatGroq(model="openai/gpt-oss-20b")
search=GoogleSerperAPIWrapper()
```

```python
agent=create_agent(
    model=llm,
    tools=[search.run],
    system_prompt="You are a Agent and can search any question on google",
    checkpointer=MemorySaver()
)
```

## 3. Invoke the search agent

Web search makes results time-dependent: rerunning the same question later can return different pages. Try asking a specific question with a clear time or source scope, and distinguish a search result snippet from a verified claim on the page.

**Checkpoint:** Why might this notebook return a different answer next week?

```python
question="total runs"
```

```python
response=agent.invoke({'messages':[{"role":"user","content":question}]},
                      {"configurable":{"thread_id":"1"}},
                      )
```
