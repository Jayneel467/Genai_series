Use LangChain's Ollama integration to send a prompt to an Ollama model and read the response.

## Learning goals
- Connect a LangChain chat-model wrapper to an Ollama service.
- Invoke a model and inspect its generated content.

## Before you start
Install Ollama, start its service, and make the model named in the code available to your Ollama account/service. Install this repository's Python dependencies. No hosted-model API key is needed for a local model; cloud-hosted Ollama models may have their own account or usage requirements.

## How to work through it
Run the setup/import cell, then invoke the model. If the request fails, confirm the Ollama service is running and the configured model name is available.

## 1. Connect to an Ollama chat model

**Ollama** runs or serves models through an Ollama runtime; `ChatOllama` is LangChain’s adapter for that runtime. It is different from a hosted provider API, though a model tagged as cloud-hosted may still use a remote service. Ensure the exact model identifier in the code is available to your Ollama installation/account. The `.invoke` call waits for a response.

**Checkpoint:** What must be running or available before `ChatOllama.invoke` can succeed?

```python
from langchain_ollama import ChatOllama
llm=ChatOllama(model="gemma4:31b-cloud")
```

```python
res=llm.invoke("I am Alex Example")
res.content
```
